# Passo 1 — Web Scraping: Dados Climáticos de Chicago (nov/2017)

**Objetivo:** Coletar a tabela de dados climáticos da página HTML e transformá-la em um DataFrame pandas chamado `weather_records`.

**URL:** https://practicum-content.s3.us-west-1.amazonaws.com/data-analyst-eng/moved_chicago_weather_2017.html

## Passos

**Passo 1** — Importação das bibliotecas: `requests`, `BeautifulSoup` e `pandas`

**Passo 2** — Requisição HTTP à URL e verificação do status code (200)

**Passo 3** — Parse do HTML com `BeautifulSoup` e localização da tabela por `attrs={"id": "weather_records"}`

**Passo 4** — Extração dos cabeçalhos (`<th>`) e das linhas de dados (`<td>`)

**Passo 5** — Criação do DataFrame `weather_records` e exibição completa

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

URL = 'https://practicum-content.s3.us-west-1.amazonaws.com/data-analyst-eng/moved_chicago_weather_2017.html'
req = requests.get(URL)
print(req.status_code)

soup = BeautifulSoup(req.text, 'html.parser')
table = soup.find('table', attrs={"id": "weather_records"})

list_cabecalho = []
for th in table.find_all('th'):
    list_cabecalho.append(th.text.strip())

list_linhas = []
for tr in table.find_all('tr'):
    celulas = tr.find_all('td')
    if len(celulas) > 0:
        list_linhas.append([td.text.strip() for td in celulas])

weather_records = pd.DataFrame(list_linhas, columns=list_cabecalho)
print(weather_records)
```
